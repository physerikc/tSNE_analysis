**This function 'create_base_legend' ensures that, regardless of which month you're looking at, the glitch colors are consistent and that the system is prepared to detect new types of noise.**

```python
def create_base_legend(glit):

    # 1 Clean the glitches data 
    """ Nesta etapa, limpamos os dados brutos para nos concentrarmos nos glitches que realmente importam. """
    glit = glit.drop(glit[glit['label'].isin(['None_of_the_Above', 'No_Glitch'])].index)
    glit = glit[glit['confidence'] >= 0.9]
    top10 = glit['label'].value_counts().head(10).index
    glit = glit[glit['label'].isin(top10)].copy()
    
    glit = glit.sort_values(by=[glitcols[0]]).reset_index(drop=True)
    clustered = glit.copy()

    # 2 Pick the 10 most common ones from the entire O3a
    """ Aqui a função olha para o conjunto de dados processados e define quem são os 10 glitches que mais aparecem em todo o período da O3a. """
    contagem_geral = Counter(clustered['label'].values)
    top_10_nomes_frequencia = [item[0] for item in contagem_geral.most_common(10)]
    
    # 3 Reorder alphabetically (the before plots are ordered)
    """ Ao ordenarmos o Top 10 alfabeticamente (sorted), garantimos que a atribuição de cores não mude se a frequência dos glitches oscilar entre uma execução e outra.
    Isso evita que, em uma rodada, o 'Blip' seja azul e, na outra, o 'Tomte' tome o lugar do azul apenas por ter tido eventos a mais. """
    top_10_alfabetico = sorted(top_10_nomes_frequencia)
    
    # 4 Create the fixed color dictionary using 'tab10'
    """ Aqui criamos o dicionário 'cor_fixa_por_glitch', onde cada nome de glitch do Top 10 ganha uma cor permanente. 
    É por causa disso que os plots agora mantêm a identidade visual em todos os meses. """
    cmap_base = plt.colormaps["tab10"]
    cor_fixa_por_glitch = {nome: cmap_base(i) for i, nome in enumerate(top_10_alfabetico)}

    print(f'{top_10_alfabetico}')
    cmap_base

    # 5 Palette for new possible glitches, using 'Set 3'
    """ Nesta parte a gente prepara o código para glitches que não estão no Top 10 geral, mas que podem aparecer mais frequentemente em um mês específico. """
    todos_os_nomes_csv = df_gspy['label'].unique()
    cmap_extras = plt.colormaps["Set3"]
    idx_extra = 0

    """ Abaixo, o código percorre todos os nomes únicos do CSV original. Se encontrar um glitch que não está no dicionário (i.e., não é do Top 10), 
    ele atribui uma cor extra do mapa 'Set3'. A parte 'idx_extra % 12' é para garantir que, se houver mais de 12 glitches extras (tamanho do 'Set3'), 
    o código não quebre e comece a repetir as cores da paleta 'Set3'. """
    for g in sorted(todos_os_nomes_csv):
        if g not in cor_fixa_por_glitch:
            cor_fixa_por_glitch[g] = cmap_extras(idx_extra % 12)
            idx_extra += 1
    cmap_extras

    return cmap_base, cmap_extras
```

**This function 'plot_the_month' is designed to create individual subplots or integrate a grid of t-SNE plots, always focusing on the 10 most frequent glitches types for each month.**

```python
def plot_the_month(proj_month, lab_month, month_name, ax=None):
    
    # lab_array = lab_month.values.ravel() if hasattr(lab_month, 'values') else np.array(lab_month).ravel()
    lab_array = np.array(lab_month).ravel()
    
    # 1 Use this month's top 10
    """ Utiliza a classe 'Counter' para contar as ocorrências de cada tipo de glitch. O método 'most_common(10)' identifica os 10 nomes de glitches que mais apareceram, salvando-os na lista top_10_mes. """
    contagem_mes = Counter(lab_array)
    top_10_mes = [item[0] for item in contagem_mes.most_common(10)]
    
    # 2 Mask for plotting only the Top 10 of the month
    """ Cria um filtro (mask) que é 'True' apenas para os pontos que pertencem a um dos 10 glitches identificados antes. Aplicamos esse filtro 
    tanto nas coordenadas espaciais (proj_plot) quanto nos rótulos (lab_plot), garantindo que apenas os dados do Top 10 sejam enviados para o gráfico. """
    mask = np.isin(lab_array, top_10_mes)
    proj_vals = proj_month.values if hasattr(proj_month, 'values') else proj_month
    proj_plot = proj_vals[mask]
    lab_plot = lab_array[mask]

    # 3 If we don't pass an axis (ax), we create a new figure.
    """ Este trecho decide se o gráfico será um arquivo único ou parte de uma grade (subplots). Se você chamar a função sem passar um eixo (ax), ela faz um gráfico isolado. """
    if ax is None:
        fig, ax = plt.subplots(figsize=(10, 6))
        is_standalone = True
    else:
        is_standalone = False
            
    # 4 Plot
    """ Aqui o código busca a cor de cada ponto em um dicionário externo chamado 'cor_fixa_por_glitch'. Isso é importante para que, 
    por exemplo, o 'Blip' tenha a mesma cor em todos os meses, permitindo a comparação visual com outros plots. """
    cores_pontos = [cor_fixa_por_glitch[g] for g in lab_plot]
    for g in top_10_mes:
        if g not in cor_fixa_por_glitch:
            cor_fixa_por_glitch[g] = cmap_extras(proximas_cores_idx % 12)
            proximas_cores_idx += 1
    
    ax.scatter(proj_plot[:, 0], proj_plot[:, 1], s=1, c=cores_pontos, alpha=0.7)
    
    # 5 Legend organized by monthly frequency
    """ Handles: em vez de deixar o Matplotlib gerar a legenda automaticamente, criamos manualmente círculos coloridos (mlines.Line2D). """
    m_size = 8 if is_standalone else 4
    handles = [mlines.Line2D([], [], color=cor_fixa_por_glitch[g], marker='o', ls='', label=f"{g} (n={contagem_mes[g]})", markersize=m_size) for g in top_10_mes]

    """ Quantidade: o rótulo da legenda (label) inclui não apenas o nome do glitch, mas também o número exato de ocorrências naquele mês: f"{g} (n={contagem_mes[g]}) """
    ax.legend(handles=handles, title="Top 10", loc='upper left', bbox_to_anchor=(0.94, 0.5), fontsize='x-small' if not is_standalone else 'medium')
    
    ax.set_title(f'Month: {month_name}')
    
    # If it's individual, it completes the plot
    if is_standalone:
        plt.tight_layout()
        plt.show()
```
