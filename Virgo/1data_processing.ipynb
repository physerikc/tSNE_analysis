```python
import os
import uproot
import glob
import pandas as pd
```

### Process the trigggers (.root) and glitches (.csv) and save as .csv

You need to run these only once. If you already created the triggers and glitches in **.csv** files, you don't need to run this code.

Triggers (.root) ~ we will use "uproot" to read the files. We are using data from **zomicron** file and save them together in **triggers_total.csv**

/////////////////////////// How to get .root files from omicron?

```python
folder = "/home/esanches/Projetos/tSNE/Virgo/zomicron"
exit_csv = "/home/esanches/Projetos/tSNE/Virgo/triggers_total.csv"

files = [f for f in os.listdir(folder) if f.endswith(".root")]

first = True

for arq in files:
    print(f"Reading: {arq}")
    caminho = os.path.join(folder, arq)
    
    with uproot.open(caminho) as f:
        tree = f["triggers;1"]
        df = tree.arrays(["time", "frequency", "snr"], library="pd")
        
    df.to_csv(exit_csv, mode="w" if first else "a", header=first, index=False)

    first = False
    del df
```

Glitches (.csv) ~ we will use "glob" to read the files. We are using data from **our_clustering** and save them together in **gdchar.csv**

/////////////////////////// How to get .csv files from our_clustering or gravity_spy clustering?

```python
pasta = "/home/esanches/Projetos/tSNE/Virgo/our_clustering/"
files = glob.glob(os.path.join(pasta, "Clustered_*.csv"))

saida_csv = "/home/esanches/Projetos/tSNE/Virgo/gdchar.csv"
first = True   # write the header only one time

for f in files:
    print(f"Reading: {f}")
    
    df = pd.read_csv(f, header=0)
    df.to_csv(saida_csv, mode='w' if first else 'a', header=first, index=False)
    
    first = False
    del df
```
