```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.manifold import TSNE
from sklearn.preprocessing import LabelEncoder
```

```python
clust = 'gspy'
run = 'O3b'

arq_name = f'{clust}{run}'
```

```python
base_data = '/home/esanches/Projetos/tSNE/LIGO'
repository = f'/home/esanches/Projetos/tSNE/LIGO/repository/{clust}/{run}'

filterr = True
```

### tSNE plots for O3b

```python
lista_dt = [2.0]
lista_perp = [10, 50, 100, 250]

fig, axs = plt.subplots(len(lista_dt), len(lista_perp), figsize=(4*len(lista_perp), 4*len(lista_dt)))

le = LabelEncoder()

for i, dt in enumerate(lista_dt):

    if filterr:
        final_vecs = np.load(repository + '/' + 'filtered_vecs' + str(dt) + '.npy')
        final_labs = np.load(repository + '/' + 'filtered_labs' + str(dt) + '.npy')
        labels = le.fit_transform(final_labs.ravel())
    else:
        final_vecs = np.load(repository + '/' + 'non_filtered_vecs' + str(dt) + '.npy')
        final_labs = np.load(repository + '/' + 'non_filtered_labs' + str(dt) + '.npy')
        labels = le.fit_transform(final_labs.ravel())

    for j, perp in enumerate(lista_perp):
        
        tsne = TSNE(n_components=2, n_jobs=-1, verbose=0, perplexity=perp, 
                    learning_rate=200, max_iter=1000, metric="euclidean", random_state=0)
        proj = tsne.fit_transform(final_vecs)
        
        ax = axs[j]
        sc = ax.scatter(proj[:, 0], proj[:, 1], s=0.3, c=labels, cmap="tab10")
        ax.set_title(f"dt: {deltat} | Perp: {perp}", fontsize=10, fontweight='bold')
        ax.tick_params(axis='both', which='major', labelsize=8)
        ax.grid(True, linestyle="--", alpha=0.3)
        
        ax.set_ylabel(f"dt = {deltat} (y axis)", fontsize=9)
        if i == len(lista_dt) - 1:
            ax.set_xlabel(f"Perp = {perp} (x axis)", fontsize=9)

plt.tight_layout()
fig.legend(handles=sc.legend_elements()[0], labels=list(le.classes_), title="Glitches", loc='center right', bbox_to_anchor=(1.1, 0.5))
plt.show()
```

```python
deltat, perp = 2.0, 100
```

```python
final_vecs = np.load(repository + '/' + 'filtered_vecs' + str(deltat) + '.npy')
final_labs = np.load(repository + '/' + 'filtered_labs' + str(deltat) + '.npy')

le = LabelEncoder()
labels = le.fit_transform(final_labs.ravel())

tsne = TSNE(n_components=2, n_jobs=-1, verbose=0, perplexity=perp, 
            learning_rate=200, max_iter=1000, metric="euclidean", random_state=0)
proj = tsne.fit_transform(final_vecs)
```

```python
plt.figure(figsize=(7, 4))

sc = plt.scatter(proj[:, 0], proj[:, 1], s=0.3, c=labels, cmap="tab10")

plt.title(f"dt={deltat} and perp={perp}")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")

plt.legend(handles=sc.legend_elements()[0], labels=list(le.classes_), 
          title="Glitches", loc='center left', bbox_to_anchor=(1.02, 0.5))

plt.tight_layout()
plt.show()
```

and for glitches

```python
glit = pd.read_csv(gdchar_source + 'gspy' + run + '.csv', usecols=glitcols)

glit = glit.drop(glit[glit['label'].isin(['None_of_the_Above', 'No_Glitch'])].index)
glit = glit[glit['confidence'] >= 0.9]

top10 = glit['label'].value_counts().head(10).index
glit = glit[glit['label'].isin(top10)].copy()

glit = glit.sort_values(by=[glitcols[0]]).reset_index(drop=True)

if filter1000 == True:
    limited_glit = glit.groupby('label', group_keys=False).apply(lambda x: x.sample(n=min(len(x), 1000), random_state=42))

    limited_glit = limited_glit.sort_values(by=[glitcols[0]]).reset_index(drop=True)
    clustered = limited_glit.copy()
else:
    clustered = glit.copy()
```
