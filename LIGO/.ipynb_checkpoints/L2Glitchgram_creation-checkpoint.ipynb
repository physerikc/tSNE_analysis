## Create the vectors from glitchgrams

```python
def glitch_to_glitchgram(unclustered, clustered, deltat, fmin=3, fmax=8192, tbins=41, fbins=30, cols=None):

    col_config = {
            'time_un': 'time',
            'freq_un': 'frequency',
            'snr_un': 'snr',
            'time_cl': 'GPStime',
            'label_cl': 'label'
    }
    if cols is not None:
        col_config.update(cols)
    
    if not unclustered[col_config['time_un']].is_monotonic_increasing:
            unclustered = unclustered.sort_values(by=col_config['time_un'])

    times_all = unclustered[col_config['time_un']].values
    freqs_all = unclustered[col_config['freq_un']].values
    snrs_all  = unclustered[col_config['snr_un']].values

    cl_times  = clustered[col_config['time_cl']].values
    cl_labels = clustered[col_config['label_cl']].values

    vecs, labs, GPSs = [], [], []

    time_edges = np.linspace(-deltat/2, deltat/2, tbins+1)
    freq_edges = np.logspace(np.log10(fmin), np.log10(fmax), fbins+1)

    for row in clustered.itertuples(index=False):
        t_center  = getattr(row, col_config['time_cl'])
        label_val = getattr(row, col_config['label_cl'])
    
        tmin = t_center - deltat/2
        tmax = t_center + deltat/2
    
        i0 = np.searchsorted(times_all, tmin, side="left")
        i1 = np.searchsorted(times_all, tmax, side="right")
    
        times = times_all[i0:i1] - t_center
        freqs = freqs_all[i0:i1]
        snrs  = snrs_all[i0:i1]

        if len(times) < 20:
            continue
    
        df_triggers = pd.DataFrame({'time': times,'frequency': freqs,'snr': snrs})
        
        t_idx = np.digitize(times, time_edges) - 1
        f_idx = np.digitize(freqs, freq_edges) - 1
            
        # initial empty matrix
        A = np.zeros((fbins, tbins), dtype=float)
    
        for ti, fi, si in zip(t_idx, f_idx, snrs):
            if 0 <= ti < tbins and 0 <= fi < fbins:
                if si > A[fi, ti]:
                    A[fi, ti] = si
                    
        '''# normalização do SNR 
        if len(snrs) > 0:
            snr_min = snrs.min()
            snr_max = snrs.max()
            if snr_max > snr_min:
                norm_snrs = (snrs - snr_min) / (snr_max - snr_min)
            else:
                norm_snrs = np.zeros_like(snrs)
        else:
            norm_snrs = snrs'''
        
        vecs.append(A.flatten(order="C"))
        labs.append(label_val)
        GPSs.append(t_center)

    vecs_final = np.array(vecs, dtype=np.float32)
    labs_final = np.array(labs)
    GPSs_final = np.array(GPSs, dtype=np.float32)
    
    return vecs_final, labs_final, GPSs_final
```

```python
unclustered = pd.read_pickle(base_data + '/' + 'gspy_O3b_unclustered_triggers.pkl')
clustered = pd.read_csv(repository + '/' + 'non_filtered_clustered.csv')
```
