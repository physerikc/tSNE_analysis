## Process trigggers

Processing the triggers: we are using data from **zomicron** file and save them together in **triggers_total.csv**

### if .root files

```python
folder = "/home/esanches/Projetos/tSNE/Virgo/zomicron"
exit_csv = "/home/esanches/Projetos/tSNE/Virgo/V1_triggers_total.csv"
```

```python
files = [f for f in os.listdir(folder) if f.endswith(".root")]

first = True

for arq in files:
    print(f"Reading: {arq}")
    caminho = os.path.join(folder, arq)
    
    with uproot.open(caminho) as f:
        tree = f["triggers;1"]
        df = tree.arrays(["time", "frequency", "snr"], library="pd")
        
    df.to_csv(exit_csv, mode="w" if first else "a", header=first, index=False)

    first = False
    del df
```
